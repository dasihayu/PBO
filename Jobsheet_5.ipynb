**Praktikum 1 - Polimorfisme dengan Inheritance dan Overriding**

```python
class Burung:
  def __init__(self,nama):
    self.nama = nama

  def terbang(self):
    print(f"{self.nama} terbang di langit langit")

  def bersuara(self):
    print(f"{self.nama} mengeluarkan suara")

class Elang(Burung):
  def __init__(self, nama, rentang_sayap):
    self.rentang_sayap = rentang_sayap
    super().__init__(nama)
  
  def terbang(self):
    print(f"{self.nama} terbang mencari mangsa")

  def bersuara(self):
    print(f"{self.nama} keras suaranya")

class Pipit(Burung):
  def __init__(self, nama, warna):
    self.warna = warna
    super().__init__(nama)
  
  def terbang(self):
    print(f"{self.nama} terbang diatas rumah")

  def bersuara(self):
    print(f"{self.nama} suaranya cit cit cuit")

def demonstrasi_aksi_burung(daftar_burung):
  print("\n Aksi Para Burung: ")
  
  for burung in daftar_burung:
    print(f"-- Aksi {burung.nama} --")
    burung.terbang()
    burung.bersuara()
    print("-" * 15)
  
if __name__ == "__main__":
  elang_jawa = Elang("Elang Jawa", 1.5)
  pipit_gereja = Pipit("Pipit Gereja", "Cokelat")
  burung_misterius = Burung("Burung Misterius")

  koleksi_burung = [elang_jawa, pipit_gereja, burung_misterius]

  demonstrasi_aksi_burung(koleksi_burung)
```

**Praktikum 2 - Polimorfisme dengan Duck Typing**

```python
class Kucing:
    def bersuara(self):
        print("Kucing: Meow!")


class Anjing:
    def bersuara(self):
        print("Anjing: Guk guk!")


class Bebek:
    def bersuara(self):
        print("Bebek: Kwek kwek!")


class Mobil:
    def jalankan(self):
        print("Mobil: Brummm!")

def buat_suara(objek_yang_bisa_bersuara):
    try:
        objek_yang_bisa_bersuara.bersuara()
    except AttributeError:
        print(f"Error: Objek dari tipe "
              f"'{type(objek_yang_bisa_bersuara).__name__}' "
              f"tidak memiliki metode 'bersuara()'")

if __name__ == "__main__":
    kucing = Kucing()
    anjing = Anjing()
    bebek = Bebek()
    mobil = Mobil()

    print("--- Demonstrasi Duck Typing ---")

    buat_suara(kucing)
    buat_suara(anjing)
    buat_suara(bebek)

    print("\nMencoba pada objek yang tidak sesuai:")
    buat_suara(mobil)
```

**Praktikum 3 - Simulasi Overloading dalam  Metode Kelas dengan *args**

```python
class KalkulatorSederhana:
    def __init__(self, nama="Kalkulator"):
        self.nama = nama
        print(f"{self.nama} siap digunakan.")

    def tambah(self, *args):
        """
        """

        print(f"\nMemanggil metode tambah dengan argumen: {args}")

        if not args:
            print("Tidak ada angka untuk dijumlahkan.")
            return 0

        total = 0
        valid_input = True

        for angka in args:
            if isinstance(angka, (int, float)):
                total += angka
            else:
                print(f"Peringatan: Argumen '{angka}' bukan angka dan diabaikan.")
                valid_input = False

        if valid_input:
            print(f"Hasil penjumlahan: {total}")
        else:
            print(f"Hasil penjumlahan (dengan beberapa input diabaikan): {total}")

        return total


if __name__ == "__main__":
    calc = KalkulatorSederhana("Calc-01")

    print("\n--- Percobaan Penjumlahan ---")

    calc.tambah(5, 10)
    calc.tambah(2, 3, 5, 10)
    calc.tambah(100)
    calc.tambah()
    calc.tambah(1, 2, "tiga", 4, 5.5)
```

**Praktikum 4 - Simulasi Overloading dalam  Metode Kelas dengan *args dan **kwargs**

```python
class DataProcessor:
    def __init__(self, processor_id):
        self.processor_id = processor_id
        print(f"DataProcessor '{self.processor_id}' diinisialisasi.")

    def process(self, *args, **kwargs):
        """
        """

        print(f"\n--- Memulai Proses pada Processor: {self.processor_id} ---")

        if args:
            print("Argumen Posisi (*args):")
            for i, arg in enumerate(args, start=1):
                print(f"  Argumen {i}: {arg}")
        else:
            print("Tidak ada argumen posisi.")

        if kwargs:
            print("Argumen Kata Kunci (**kwargs):")
            for key, value in kwargs.items():
                print(f"  {key}: {value}")
        else:
            print("Tidak ada argumen kata kunci.")

        print("--- Proses Selesai ---")

if __name__ == "__main__":
    processor = DataProcessor("DP-001")

    print("\n=== Percobaan 1 ===")
    processor.process(
        "Data Sensor",
        "Suhu",
        25.5,
        lokasi="Lab Komputer",
        status="Aktif"
    )

    print("\n=== Percobaan 2 ===")
    processor.process(
        100,
        200,
        300
    )

    print("\n=== Percobaan 3 ===")
    processor.process(
        nama="Andi",
        umur=21,
        jurusan="Teknik Informatika"
    )

    print("\n=== Percobaan 4 ===")
    processor.process()
```

**Prakikum 5 - Kombinasi Polimorfisme  (Inheritance & Duck Typing)**

```python
from abc import ABC, abstractmethod
import math


class Bentuk(ABC):

    @abstractmethod
    def hitung_luas(self):
        pass

    def nama_bentuk(self):
        return "Bentuk Umum"


class Lingkaran(Bentuk):
    def __init__(self, radius):
        self.radius = radius

    def hitung_luas(self):
        return math.pi * self.radius * self.radius

    def nama_bentuk(self):
        return "Lingkaran"


class Persegi(Bentuk):
    def __init__(self, sisi):
        self.sisi = sisi

    def hitung_luas(self):
        return self.sisi * self.sisi

    def nama_bentuk(self):
        return "Persegi"


class TaplakMeja:
    def __init__(self, panjang, lebar):
        self.panjang = panjang
        self.lebar = lebar

    def hitung_luas(self):
        return self.panjang * self.lebar

    def deskripsi(self):
        return "Taplak meja persegi panjang"


def tampilkan_info_luas(objek):
    try:
        print(f"\nObjek: {objek.__class__.__name__}")

        if hasattr(objek, "nama_bentuk"):
            print(f"Nama Bentuk: {objek.nama_bentuk()}")

        luas = objek.hitung_luas()
        print(f"Luas: {luas}")

    except AttributeError:
        print(f"Error: Objek '{objek.__class__.__name__}' "
              f"tidak memiliki metode yang diperlukan.")

    except TypeError:
        print(f"Error: Metode belum diimplementasikan.")

if __name__ == "__main__":

    lingkaran = Lingkaran(7)
    persegi = Persegi(5)
    taplak = TaplakMeja(4, 6)

    print("=== Demonstrasi Polimorfisme ===")

    tampilkan_info_luas(lingkaran)
    tampilkan_info_luas(persegi)
    tampilkan_info_luas(taplak)

    print("\n=== Percobaan Error ===")

    tampilkan_info_luas("Hello World")
```

**Praktikum 6 - Kontrol Perilaku Polimorfik  dengan isinstance()**

```python
class Burung:
    def __init__(self, nama):
        self.nama = nama

    def terbang(self):
        print(f"{self.nama} terbang dengan cara umum.")

    def bersuara(self):
        print(f"{self.nama} mengeluarkan suara burung.")


class Elang(Burung):
    def __init__(self, nama, rentang_sayap):
        super().__init__(nama)
        self.rentang_sayap = rentang_sayap

    def terbang(self):
        print(f"{self.nama} melayang tinggi di angkasa.")

    def bersuara(self):
        print(f"{self.nama} bersuara: Screeeech!")

    def berburu(self):
        print(f"{self.nama} sedang berburu mangsa.")


class Pipit(Burung):
    def __init__(self, nama, warna_bulu):
        super().__init__(nama)
        self.warna_bulu = warna_bulu

    def terbang(self):
        print(f"{self.nama} terbang lincah di antara pepohonan.")

    def bersuara(self):
        print(f"{self.nama} berkicau: Cit cit cit!")

    def membangun_sarang(self):
        print(f"{self.nama} sedang membangun sarang.")


def interaksi_dengan_burung(objek):
    print("\n=== Interaksi dengan Burung ===")

    if isinstance(objek, Burung):

        print(f"Nama Burung: {objek.nama}")

        objek.terbang()
        objek.bersuara()

        if isinstance(objek, Elang):
            objek.berburu()

        elif isinstance(objek, Pipit):
            objek.membangun_sarang()

    else:
        print(f"Objek '{objek}' bukan merupakan Burung.")


if __name__ == "__main__":

    elang = Elang("Elang Jawa", 2.1)
    pipit = Pipit("Pipit Sawah", "Coklat")
    burung_biasa = Burung("Burung Umum")
    angka = 100

    # Demonstrasi
    interaksi_dengan_burung(elang)
    interaksi_dengan_burung(pipit)
    interaksi_dengan_burung(burung_biasa)

    print("\n=== Percobaan Objek Non-Burung ===")
    interaksi_dengan_burung(angka)
```

**Latihan**

```python
class Komputer:
    def __init__(self, merk, processor, ram_gb):
        self.merk = merk
        self.processor = processor
        self.ram_gb = ram_gb

    def info_spesifikasi(self):
        print(f"Merk: {self.merk}")
        print(f"Processor: {self.processor}")
        print(f"RAM: {self.ram_gb} GB")

    def jalankan_aplikasi(self, nama_aplikasi):
        print(f"{self.merk} menjalankan aplikasi: {nama_aplikasi}...")

    def install_software(self, *args):

        if not args:
            print("Tidak ada software yang dipilih.")
            return

        for software in args:
            print(f"Menginstall {software} di {self.merk}...")


class Laptop(Komputer):
    def __init__(self, merk, processor, ram_gb,
                 ukuran_layar_inch, berat_kg):

        super().__init__(merk, processor, ram_gb)

        self.ukuran_layar_inch = ukuran_layar_inch
        self.berat_kg = berat_kg

    def info_spesifikasi(self):
        super().info_spesifikasi()

        print(f"Ukuran Layar: {self.ukuran_layar_inch} inch")
        print(f"Berat: {self.berat_kg} Kg")

        print("-" * 30)


class Desktop(Komputer):
    def __init__(self, merk, processor, ram_gb,
                 jenis_casing, monitor_external):

        super().__init__(merk, processor, ram_gb)

        self.jenis_casing = jenis_casing
        self.monitor_external = monitor_external

    def info_spesifikasi(self):
        super().info_spesifikasi()

        print(f"Jenis Casing: {self.jenis_casing}")
        print(f"Monitor External: "
              f"{'Ya' if self.monitor_external else 'Tidak'}")

        print("-" * 30)

def cetak_semua_spesifikasi(daftar_komputer):

    print("\n=== DAFTAR SPESIFIKASI KOMPUTER ===")

    for komputer in daftar_komputer:
        komputer.info_spesifikasi()

if __name__ == "__main__":
    laptop1 = Laptop(
        "ASUS",
        "Intel Core i7",
        16,
        15.6,
        1.8
    )

    desktop1 = Desktop(
        "Acer",
        "AMD Ryzen 5",
        32,
        "Tower",
        True
    )

    daftar_komputer = [laptop1, desktop1]

    cetak_semua_spesifikasi(daftar_komputer)

    print("\n=== MENJALANKAN APLIKASI ===")
    laptop1.jalankan_aplikasi("Visual Studio Code")

    print("\n=== INSTALL SOFTWARE ===")

    laptop1.install_software("Google Chrome")

    print()

    laptop1.install_software(
        "Microsoft Office",
        "Photoshop",
        "Discord"
    )
```
